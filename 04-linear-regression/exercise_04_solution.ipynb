### Among the following three features, find the best single feature that minimizes the MSE for linear regression. Plot the MEDV vs the best feature with the predicted line. Calculate the MSE.

- INDUS
- DIS
- PTRATIO

# 1) Load data

```python
import pandas as pd

df = pd.read_csv("Housing price dataset_modified.csv")
```

# 2) Linear regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

X = df[['INDUS']].values
y = df['MEDV'].values

slr = LinearRegression()
slr.fit(X, y)

slope_INDUS = slr.coef_[0]
intercept_INDUS = slr.intercept_

y_pred = intercept_INDUS + slope_INDUS * X

# MSE
mse_INDUS = mean_squared_error(y, y_pred)
print('MSE_INDUS:', mse_INDUS)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

X = df[['DIS']].values
y = df['MEDV'].values

slr = LinearRegression()
slr.fit(X, y)

slope_DIS = slr.coef_[0]
intercept_DIS = slr.intercept_

y_pred = intercept_DIS + slope_DIS * X

# MSE
mse_DIS = mean_squared_error(y, y_pred)
print('MSE_DIS:', mse_DIS)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

X = df[['PTRATIO']].values
y = df['MEDV'].values

slr = LinearRegression()
slr.fit(X, y)

slope_PTRATIO = slr.coef_[0]
intercept_PTRATIO = slr.intercept_

y_pred = intercept_PTRATIO + slope_PTRATIO * X

# MSE
mse_PTRATIO = mean_squared_error(y, y_pred)
print('MSE_PTRATIO:', mse_PTRATIO)
```

The best single feature among 'INDUS', 'DIS', and 'PTRATIO' is 'PTRATIO'.

```python
import matplotlib.pyplot as plt

X = df[['PTRATIO']].values
y = df['MEDV'].values

# Plot the ground truth points
plt.scatter(X, y, c='steelblue', edgecolor='white', s=70)

# Calculate the predictions
y_pred = intercept_PTRATIO + slope_PTRATIO * X

# MSE
mse = mean_squared_error(y, y_pred)
print('MSE:', mse)

# Plot the fitted (predicted) line
plt.plot(X, y_pred, color='red', lw=2, label='Fitted (predicted) line')

# Add labels
plt.xlabel('Average number of rooms [RM]')
plt.ylabel('Price in $1000s [MEDV]')
plt.legend()
plt.show()
```
