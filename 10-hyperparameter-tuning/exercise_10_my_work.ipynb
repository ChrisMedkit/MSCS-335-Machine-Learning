Based on Practice 10, find the best 'alpha' and 'polynomial degree'.

 - **'alpha'**: [1e-3, 1e-2, 1e-1, 1, 10]

 - **'polynomial degree'**: [1, 2, 3]


1) Use 'Housing price dataset_modified.csv'.

2) Use the same splitting as in Practice 10.

3) Change **'Ridge'** regression into **'Lasso'** regression.

4) With the best parameters, calculate the MSE on the test set.

5) For other simulation environments, use the same setting in Practice 10.

```python
import numpy as np
import pandas as pd

df=pd.read_csv('Housing price dataset_modified.csv')

from sklearn.model_selection import train_test_split

X=df.drop(columns="MEDV")
y=df["MEDV"]

X_train, X_test, y_train, y_test = train_test_split (
    X,y,test_size=0.2, random_state=15
)

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,PolynomialFeatures
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error

pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("poly", PolynomialFeatures(include_bias=False)),
    ("lasso", Lasso(tol=0.1))
])

param_grid = {
    "poly__degree":[1,2,3],
    "lasso__alpha":[1e-3, 1e-2,1e-1,1,10]
}

from sklearn.model_selection import GridSearchCV

num_canidates = len(param_grid["poly__degree"])*len(param_grid["lasso__alpha"])

grid=GridSearchCV(
    pipe,
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_squared_error",
    refit=True,
    return_train_score=True,
    n_jobs=None
    )

grid.fit(X_train, y_train)

print("Best parameters:", grid.best_params_)


best_degree = grid.best_params_["poly__degree"]
best_alpha = grid.best_params_["lasso__alpha"]
best_cv_mse = -grid.best_score_


best_model = grid.best_estimator_

from sklearn.metrics import mean_squared_error

y_test_pred = best_model.predict(X_test)
test_mse = mean_squared_error(y_test, y_test_pred)

print("Final chosen degree:", best_degree)
print("Final chosen alpha :", best_alpha)
print("Final test MSE     :", test_mse)
```
