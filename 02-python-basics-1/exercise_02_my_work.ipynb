## Data

```python
# Data
names = ["img_001.jpg","img_002.png","img_003.jpg"," img_004.JPG ","img_005.jpeg","","img_006.jpg"]
labels = ["cat","dog","cat","dog","dog","", "cat"]

print("names:", names)
print("labels:", labels)
```


## Step 1 — Clean & filter pairs
Rules:
1) Strip whitespace from names and labels; convert **name** to lowercase.
2) Keep only names ending with **`.jpg` or `.jpeg`**.
3) Drop any pair with an empty name or empty label.

Create:
- `cleaned_pairs`: list of `(name, label)` after cleaning/filtering
- `cleaned_names`, `cleaned_labels`: unpacked lists


```python
# TODO: build cleaned_pairs, cleaned_names, cleaned_labels
cleaned_pairs = []
for n, y in zip(names, labels):
    # Fill out here
    n =n.strip().lower()
    y =y.strip().lower()
    if not n or not y:
        continue
    if not (n.endswith(".jpg") or n.endswith(".jpeg")):
        continue
    cleaned_pairs.append((n, y))

cleaned_names = [n for n, _ in cleaned_pairs]
cleaned_labels = [y for _, y in cleaned_pairs]
print(cleaned_pairs)
```


## Step 2 — Build mappings & variants
- `label_to_idx`: alphabetical mapping of unique labels → index
- `idx_to_label`: inverse mapping
- `cleaned_pairs_idx`: list of tuples `(name, label_index)`
- `first_three`: first three names (slicing)


```python
# TODO
label_to_idx = {lab: i for i, lab in enumerate(sorted(set(cleaned_labels)))}
idx_to_label = {i: lab for lab, i in label_to_idx.items()}

# Add the index to each cleaned pair
cleaned_pairs_idx = [ (n,label_to_idx[y])for n, y in cleaned_pairs]
first_three = cleaned_names[:3]
print(label_to_idx, idx_to_label)
print(cleaned_pairs_idx)
print(first_three)
```


## Step 3 — Quick views & stats
- `dog_pos`: indices where label == `"dog"` using `enumerate(cleaned_labels, start=1)`
- `jpg_only`: names that end with `.jpg` (not `.jpeg`) using `n.endswith(".jpg")`
- `total_chars`: the number of characters in `cleaned_names`


```python
# TODO
dog_pos = [i for i, y in enumerate(cleaned_labels, start=1) if y=="dog"]
jpg_only = [n for n in cleaned_names if n.endswith(".jpg")]
total_chars = sum(len(n) for n in cleaned_names)
print(dog_pos)
print(jpg_only)
print(total_chars)
```
