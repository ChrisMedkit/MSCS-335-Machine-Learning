Use 'Housing price dataset_modified.csv'.

1) Use the following values of 'alpha': 0.01, 0.1, 1, 10, 100

2) For each alpha, train the Ridge regression and Lasso regression using the code below.

3) Record the train MSE and validation MSE in the tables below.

4) Which alpha gives the best validation MSE for Ridge?

5) Which alpha gives the best validation MSE for Lasso?

```python
import pandas as pd

# Read the CSV file uploaded to the current Colab working directory
df = pd.read_csv("Housing price dataset_modified.csv")

import numpy as np

# Showing the first 5 rows with .head()
print(df.shape)
df.head()
```

```python
from sklearn.model_selection import train_test_split

# Use all features except MEDV as input
X = df.drop(columns=["MEDV"]).values
y = df["MEDV"].values

X_train, X_val, y_train, y_val = train_test_split(
    X, y, train_size=100, test_size=100, random_state=42
)

print("X_train shape:", X_train.shape)
print("X_val shape :", X_val.shape)
print("y_train shape:", y_train.shape)
print("y_val shape :", y_val.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Standardize the features using the training set only
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error

# Your code here

alpha_list = [0.01, 0.1, 1, 10, 100]
results_ridge = []
results_lasso = []

for alpha in alpha_list:
    # Your code here
    print("alpha:", alpha)

    ridge = Ridge(alpha=alpha)
    lasso = Lasso(alpha=alpha)

    ridge.fit(X_train_scaled, y_train)
    lasso.fit(X_train_scaled, y_train)

    y_train_pred_ridge = ridge.predict(X_train_scaled)
    y_val_pred_ridge = ridge.predict(X_val_scaled)
    train_mse_ridge = mean_squared_error(y_train, y_train_pred_ridge)
    val_mse_ridge = mean_squared_error(y_val, y_val_pred_ridge)

    y_train_pred_lasso = lasso.predict(X_train_scaled)
    y_val_pred_lasso = lasso.predict(X_val_scaled)
    train_mse_lasso = mean_squared_error(y_train, y_train_pred_lasso)
    val_mse_lasso = mean_squared_error(y_val, y_val_pred_lasso)

    print("\nRidge Regression")
    print("Train MSE:", mean_squared_error(y_train, y_train_pred_ridge))
    print("Val  MSE:", mean_squared_error(y_val, y_val_pred_ridge))

    print("\nLasso Regression")
    print("Train MSE:", mean_squared_error(y_train, y_train_pred_lasso))
    print("Val  MSE:", mean_squared_error(y_val, y_val_pred_lasso))
    print("\n")

    results_ridge.append({
        "alpha": alpha,
        "Training MSE": train_mse_ridge,
        "Validation MSE": val_mse_ridge,
    })

    results_lasso.append({
        "alpha": alpha,
        "Training MSE": train_mse_lasso,
        "Validation MSE": val_mse_lasso,
    })
```
