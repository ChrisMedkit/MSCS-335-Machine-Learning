2) Predict 'MEDV' using both 'RM' and 'INDUS' with polynomial regression of order 2.

3) Calculate the MSE.

4) Plot a 3D graph of scattered data and regression plane using 'plotly.graph_objects' as in Practice 5.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import PolynomialFeatures
import plotly.graph_objects as go


df = pd.read_csv("Housing price dataset_modified.csv")

print(df.shape)
df.head()

#2
X = df[['RM', 'INDUS']]
y = df['MEDV']

poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)

reg = LinearRegression()
reg.fit(X_poly, y)

y_pred = reg.predict(X_poly)

#3
mse = mean_squared_error(y, y_pred)
print(f"Mean Squared Error: {mse}")


#4
x_axis = X.iloc[:, 0]   # first feature
y_axis = X.iloc[:, 1]   # second feature
z_axis = y         # target

x_pred = np.linspace(x_axis.min(), x_axis.max(), 30)
y_pred = np.linspace(y_axis.min(), y_axis.max(), 30)

xx_pred, yy_pred = np.meshgrid(x_pred, y_pred)

# It's important to transform model_viz with the same PolynomialFeatures object
# that was used to train the regression model.
# Otherwise, the model_viz will not have the correct number of features for prediction
# if the polynomial features are of degree > 1.
# First, create a DataFrame for model_viz with the original column names
model_viz_df = pd.DataFrame(np.array([xx_pred.flatten(), yy_pred.flatten()]).T, columns=['RM', 'INDUS'])
model_viz_poly = poly.transform(model_viz_df)

z_pred = reg.predict(model_viz_poly)
zz_pred = z_pred.reshape(xx_pred.shape)

fig = go.Figure()

fig.add_trace(go.Scatter3d(
    x=x_axis,
    y=y_axis,
    z=z_axis,
    mode='markers',
    marker=dict(
        size=4,
        color='black',
        opacity=0.5
    ),
    name='Data points'
))

fig.add_trace(go.Surface(
    x=xx_pred,
    y=yy_pred,
    z=zz_pred,
    opacity=0.7,
    showscale=False,
    name='Regression surface'
))

fig.update_layout(
    width=800,
    height=700,
    scene=dict(
        xaxis_title='INDUS',
        yaxis_title='RM',
        zaxis_title='MEDV'
    ),
    margin=dict(l=0, r=0, b=0, t=40),
    title='3D Regression Plane'
)

fig.show()
```
