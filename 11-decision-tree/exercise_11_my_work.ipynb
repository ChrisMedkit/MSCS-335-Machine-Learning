```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine

# Load the Iris dataset from scikit-learn
# as_frame=True returns pandas objects, which are easier to inspect and visualize
wine = load_wine(as_frame=True)

# X contains the input features (4 features)
X = wine.data

# y contains the class labels (0, 1, 2)
y = wine.target

# Feature names and class names
feature_names = wine.feature_names
class_names   = wine.target_names

print("Feature names:", feature_names)
print("Class names  :", class_names)
print("X shape      :", X.shape)
print("y shape      :", y.shape)


label_to_name = {i: name for i, name in enumerate(class_names)}
y_names = y.map(label_to_name)

print("Class counts:")
print(y_names.value_counts())


from sklearn.model_selection import train_test_split

# Split the data with stratification to preserve class ratios
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=2026,
    stratify=y
)

from sklearn.tree import DecisionTreeClassifier

# Train a Decision Tree classifier
# No pipeline needed since Decision Trees do not require feature scaling
clf = DecisionTreeClassifier(
    max_depth=4,
    criterion='entropy',
    random_state=2026
)

clf.fit(X_train, y_train)

print("Training accuracy:", clf.score(X_train, y_train))
print("Test accuracy    :", clf.score(X_test, y_test))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

y_pred = clf.predict(X_test)

print("Confusion Matrix:\n")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred, target_names=class_names))
```
