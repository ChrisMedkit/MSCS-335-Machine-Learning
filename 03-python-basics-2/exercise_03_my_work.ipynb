```python
import csv
from pathlib import Path

def add_prefix(xs):
    return  [f"img_{x}" for x in xs]

data = ["a.jpg", "b.jpg", "c.jpg", "d.jpg"]

data_prefix = add_prefix(data)
print(data_prefix)
```

Output: ['img_a.jpg', 'img_b.jpg', 'img_c.jpg', 'img_d.jpg']

```python
data_prefix_idx = [(i+1, x) for i, x in enumerate(data_prefix)]
print(data_prefix_idx)
```

Output: [(1, 'img_a.jpg'), (2, 'img_b.jpg'), (3, 'img_c.jpg'), (4, 'img_d.jpg')]

```python
# Write data in CSV format
csv_path = Path("data_prefix_idx.csv")
with csv_path.open("w", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(data_prefix_idx)

records = []
with csv_path.open("r", encoding="utf-8") as f:
    reader = csv.DictReader(f, fieldnames=['No,','filename'])
    for row in reader:
        records.append(row)
    print(records)
```

Output: [{'No.': '1', 'filename': 'img_a.jpg'}, {'No.': '2', 'filename': 'img_b.jpg'}, {'No.': '3', 'filename': 'img_c.jpg'}, {'No.': '4', 'filename': 'img_d.jpg'}]
